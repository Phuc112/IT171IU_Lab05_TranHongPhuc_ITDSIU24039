# <span style="color:red">2. Theory</span>

## <span style="color:red">2.1 Why not always use all predictors?</span>

<span style="color:red">Least squares on all $p$ predictors has low bias but can have **high variance** when $p$ is not much smaller than $n$, and is impossible when $p > n$. It also keeps irrelevant variables, which hurts interpretation. Three families of alternatives:</span>

1. <span style="color:red">**Subset selection** – fit least squares on a subset of the predictors;</span>
2. <span style="color:red">**Shrinkage (regularisation)** – fit all predictors but shrink the coefficients towards zero;</span>
3. <span style="color:red">**Dimension reduction** – regress on $M < p$ linear combinations of the predictors.</span>

## <span style="color:red">2.2 Subset selection</span>

- <span style="color:red">**Best subset:** for each $k = 1,\dots,p$ fit all $\binom{p}{k}$ models and keep the best one $\mathcal M_k$ (lowest RSS) – $2^p$ models in total.</span>
- <span style="color:red">**Forward stepwise:** start from the null model and add, one at a time, the variable that improves the fit most – only $1 + p(p+1)/2$ models.</span>
- <span style="color:red">**Backward stepwise:** start from the full model ($n > p$ required) and remove variables one at a time.</span>

<span style="color:red">The final choice among $\mathcal M_0,\dots,\mathcal M_p$ must use an estimate of the **test** error, either an adjusted training error or (cross-)validation:</span>

$$\color{red}{C_p = \frac{1}{n}\left(\mathrm{RSS} + 2d\hat\sigma^2\right),\qquad \mathrm{BIC} = \frac{1}{n}\left(\mathrm{RSS} + \log(n)\,d\hat\sigma^2\right),\qquad R^2_{adj} = 1 - \frac{\mathrm{RSS}/(n-d-1)}{\mathrm{TSS}/(n-1)},}$$

<span style="color:red">where $d$ is the number of predictors and $\hat\sigma^2$ estimates the error variance. Small $C_p$/BIC and large $R^2_{adj}$ are good; BIC penalises large models more heavily.</span>

## <span style="color:red">2.3 Ridge regression and the lasso</span>

<span style="color:red">Both add a penalty to the residual sum of squares:</span>

$$\color{red}{\text{ridge:}\quad \min_\beta\ \mathrm{RSS} + \lambda\sum_{j=1}^p \beta_j^2, \qquad\qquad \text{lasso:}\quad \min_\beta\ \mathrm{RSS} + \lambda\sum_{j=1}^p |\beta_j| .}$$

- <span style="color:red">$\lambda = 0$ gives least squares; $\lambda\to\infty$ shrinks all slopes to zero. The intercept is not penalised.</span>
- <span style="color:red">Ridge shrinks coefficients **towards** zero but keeps all variables; the lasso sets some coefficients **exactly** to zero – it performs *variable selection* and gives sparse, interpretable models.</span>
- <span style="color:red">The penalties depend on the scale of the predictors, so we **standardise** them first.</span>
- <span style="color:red">$\lambda$ is chosen by cross-validation. The **one-standard-error rule** picks the largest $\lambda$ whose CV error is within one standard error of the minimum (a simpler model that is "as good").</span>

<span style="color:red">In scikit-learn, `ElasticNet(alpha=λ, l1_ratio=0)` is ridge and `l1_ratio=1` is the lasso; scikit-learn divides the RSS by $2n$, so its `alpha` is on a different scale than the $\lambda$ of the textbook.</span>

## <span style="color:red">2.4 Dimension-reduction methods</span>

<span style="color:red">Construct $M < p$ linear combinations $Z_m = \sum_j \phi_{jm}X_j$ and regress $Y$ on $Z_1,\dots,Z_M$:</span>

- <span style="color:red">**PCR** uses the first $M$ **principal components** of $X$ – directions of largest variance, found *without* looking at $Y$ (unsupervised).</span>
- <span style="color:red">**PLS** chooses directions that have high variance **and** high correlation with $Y$ (supervised).</span>

<span style="color:red">$M$ is chosen by cross-validation. PCR performs well when a few components capture most of the variation in $X$ that is related to $Y$.</span>

## <span style="color:red">2.5 High-dimensional data ($p \approx n$ or $p > n$)</span>

<span style="color:red">With many predictors, least squares interpolates the training data ($R^2 = 1$, training MSE $= 0$) yet predicts badly; $C_p$, AIC and BIC break down because $\hat\sigma^2$ cannot be estimated. Regularisation, a careful choice of tuning parameters on independent data, and modest claims about which variables "matter" become essential.</span>

## 3.1 Subset Selection Methods

### Forward Selection

We will  apply the forward-selection approach to the  `Hitters`
data.  We wish to predict a baseball player’s `Salary` on the
basis of various statistics associated with performance in the
previous year.

First of all, we note that the `Salary` variable is missing for
some of the players.  The `np.isnan()`  function can be used to
identify the missing observations. It returns an array
of the same shape as the input vector, with a `True` for any elements that
are missing, and a `False` for non-missing elements.  The
`sum()`  method can then be used to count all of the
missing elements.

````python
Hitters = load_data('Hitters')
np.isnan(Hitters['Salary']).sum()
````

 We see that `Salary` is missing for 59 players. The
`dropna()`  method of data frames removes all of the rows that have missing
values in any variable (by default --- see  `Hitters.dropna?`).

````python
Hitters = Hitters.dropna();
Hitters.shape
````

We first choose the best model using forward selection based on $C_p$ (6.2). This score
is not built in as a metric to `sklearn`. We therefore define a function to compute it ourselves, and use
it as a scorer. By default, `sklearn` tries to maximize a score, hence
  our scoring function  computes the negative $C_p$ statistic.

````python
def nCp(sigma2, estimator, X, Y):
    "Negative Cp statistic"
    n, p = X.shape
    Yhat = estimator.predict(X)
    RSS = np.sum((Y - Yhat)**2)
    return -(RSS + 2 * p * sigma2) / n 
````

We need to estimate the residual variance $\sigma^2$, which is the first argument in our scoring function above.
We will fit the biggest model, using all the variables, and estimate $\sigma^2$ based on its MSE.

````python
design = MS(Hitters.columns.drop('Salary')).fit(Hitters)
Y = np.array(Hitters['Salary'])
X = design.transform(Hitters)
sigma2 = OLS(Y,X).fit().scale
````

The function `sklearn_selected()` expects a scorer with just three arguments --- the last three in the definition of `nCp()` above. We use the function `partial()` first seen in Section 5.3.3 to freeze the first argument with our estimate of $\sigma^2$.

````python
neg_Cp = partial(nCp, sigma2)
````

We can now use `neg_Cp()` as a scorer for model selection.

Along with a score we need to specify the search strategy. This is done through the object
`Stepwise()`  in the `ISLP.models` package. The method `Stepwise.first_peak()`
runs forward stepwise until any further additions to the model do not result
in an improvement in the evaluation score. Similarly, the method `Stepwise.fixed_steps()`
runs a fixed number of steps of stepwise search.

````python
strategy = Stepwise.first_peak(design,
                               direction='forward',
                               max_terms=len(design.terms))
````

We now fit a linear regression model with `Salary` as outcome using forward
selection. To do so, we use the function `sklearn_selected()`  from the `ISLP.models` package. This takes
a model from `statsmodels` along with a search strategy and selects a model with its
`fit` method. Without specifying a `scoring` argument, the score defaults to MSE, and so all 19 variables will be
selected.

````python
hitters_MSE = sklearn_selected(OLS,
                               strategy)
hitters_MSE.fit(Hitters, Y)
hitters_MSE.selected_state_
````

Using `neg_Cp` results in a smaller model, as expected, with just 10 variables selected.

````python
hitters_Cp = sklearn_selected(OLS,
                               strategy,
                               scoring=neg_Cp)
hitters_Cp.fit(Hitters, Y)
hitters_Cp.selected_state_
````

### Choosing Among Models Using the Validation Set Approach and Cross-Validation

As an  alternative to using $C_p$, we might try cross-validation to select a model in forward selection. For this, we need a
method that stores the full path of models found in forward selection, and allows predictions for each of these. This can be done with the `sklearn_selection_path()`
estimator from `ISLP.models`. The function `cross_val_predict()` from `ISLP.models`
computes the cross-validated predictions for each of the models
along the path, which we can use to evaluate the cross-validated MSE
along the path.

Here we define a strategy that fits the full forward selection path.
While there are various parameter choices for `sklearn_selection_path()`,
we use the defaults here, which selects the model at each step based on the biggest reduction  in RSS.

````python
strategy = Stepwise.fixed_steps(design,
                                len(design.terms),
                                direction='forward')
full_path = sklearn_selection_path(OLS, strategy)
````

We now fit the full forward-selection path on the `Hitters` data and compute the fitted values.

````python
full_path.fit(Hitters, Y)
Yhat_in = full_path.predict(Hitters)
Yhat_in.shape
````

This gives us an array of fitted values --- 20 steps in all, including the fitted mean for the null model --- which we can use to evaluate
in-sample MSE. As expected, the in-sample MSE improves each step we take,
indicating we must use either the validation or cross-validation
approach to select the number of steps. We fix the y-axis to range from
50,000 to 250,000 to compare to the cross-validation and validation
set MSE below, as well as other methods such as ridge regression, lasso and
principal components regression.

````python
mse_fig, ax = subplots(figsize=(8,8))
insample_mse = ((Yhat_in - Y[:,None])**2).mean(0)
n_steps = insample_mse.shape[0]
ax.plot(np.arange(n_steps),
        insample_mse,
        'k', # color black
        label='In-sample')
ax.set_ylabel('MSE',
              fontsize=20)
ax.set_xlabel('# steps of forward stepwise',
              fontsize=20)
ax.set_xticks(np.arange(n_steps)[::2])
ax.legend()
ax.set_ylim([50000,250000]);
````

Notice the expression `None` in `Y[:,None]` above.
This adds an axis (dimension) to the one-dimensional array `Y`,
which allows it to be recycled when subtracted from the two-dimensional `Yhat_in`.

We are now ready to use cross-validation to estimate test error along
the model path. We must use *only the training observations* to perform all aspects of model-fitting --- including
variable selection.  Therefore, the determination of which model of a
given size is best must be made using \emph{only the training
  observations} in each training fold. This point is subtle but important.  If the full data
set is used to select the best subset at each step, then the validation
set errors and cross-validation errors that we obtain will not be
accurate estimates of the test error.

We now compute the cross-validated predicted values using 5-fold cross-validation.

````python
K = 5
kfold = skm.KFold(K,
                  random_state=0,
                  shuffle=True)
Yhat_cv = skm.cross_val_predict(full_path,
                                Hitters,
                                Y,
                                cv=kfold)
Yhat_cv.shape
````

`skm.cross_val_predict()`
The prediction matrix `Yhat_cv` is the same shape as `Yhat_in`; the difference is that the predictions in each row, corresponding to a particular sample index, were made from models fit on a training fold that did not include that row.

At each model along the path, we compute the MSE in each of the cross-validation folds.
These we will average to get the  mean MSE, and can also use the individual values to compute a crude estimate of the standard error of the mean. {The estimate is crude because the five error estimates  are based on overlapping training sets, and hence are not independent.}
Hence we must know the test indices for each cross-validation
split. This can be found by using the `split()` method of `kfold`. Because
we  fixed the random state above, whenever we split any array with the same
number of rows as $Y$ we recover the same training and test indices, though we simply
ignore the training indices below.

````python
cv_mse = []
for train_idx, test_idx in kfold.split(Y):
    errors = (Yhat_cv[test_idx] - Y[test_idx,None])**2
    cv_mse.append(errors.mean(0)) # column means
cv_mse = np.array(cv_mse).T
cv_mse.shape
````

We now add the cross-validation error estimates to our MSE plot.
We include the mean error across the five folds, and the estimate of the standard error of the mean.

````python
ax.errorbar(np.arange(n_steps), 
            cv_mse.mean(1),
            cv_mse.std(1) / np.sqrt(K),
            label='Cross-validated',
            c='r') # color red
ax.set_ylim([50000,250000])
ax.legend()
mse_fig
````

To repeat the above using the validation set approach, we simply change our
`cv` argument to a validation set: one random split of the data into a test and training. We choose a test size
of 20%, similar to the size of each test set in 5-fold cross-validation.`skm.ShuffleSplit()`

````python
validation = skm.ShuffleSplit(n_splits=1, 
                              test_size=0.2,
                              random_state=0)
for train_idx, test_idx in validation.split(Y):
    full_path.fit(Hitters.iloc[train_idx],
                  Y[train_idx])
    Yhat_val = full_path.predict(Hitters.iloc[test_idx])
    errors = (Yhat_val - Y[test_idx,None])**2
    validation_mse = errors.mean(0)
````

 As for the in-sample MSE case, the validation set approach does not provide standard errors.

````python
ax.plot(np.arange(n_steps), 
        validation_mse,
        'b--', # color blue, broken line
        label='Validation')
ax.set_xticks(np.arange(n_steps)[::2])
ax.set_ylim([50000,250000])
ax.legend()
mse_fig
````

### Ridge Regression
We will use the function `skl.ElasticNet()` to fit both  ridge and the lasso.
To fit a *path* of ridge regressions models, we use
`skl.ElasticNet.path()`, which can fit both ridge and lasso, as well as a hybrid mixture;  ridge regression
corresponds to `l1_ratio=0`.
It is good practice to standardize the columns of `X` in these applications, if the variables are measured in different units. Since `skl.ElasticNet()` does no normalization, we have to take care of that ourselves.
Since we
standardize first, in order to find coefficient
estimates on the original scale, we must *unstandardize*
the coefficient estimates. The parameter
$\lambda$ in (6.5) and (6.7) is called `alphas` in `sklearn`. In order to
be consistent with the rest of this chapter, we use `lambdas`
rather than `alphas` in what follows.  {At the time of publication, ridge fits like the one in code chunk [23] issue unwarranted convergence warning messages; we suppressed these when we filtered the warnings above.}

````python
Xs = X - X.mean(0)[None,:]
X_scale = X.std(0)
Xs = Xs / X_scale[None,:]
lambdas = 10**np.linspace(8, -2, 100) / Y.std()
soln_array = skl.ElasticNet.path(Xs,
                                 Y,
                                 l1_ratio=0.,
                                 alphas=lambdas)[1]
soln_array.shape
````

Here we extract the array of coefficients corresponding to the solutions along the regularization path.
By default the `skl.ElasticNet.path` method fits a path along
an automatically selected range of $\lambda$ values, except for the case when
`l1_ratio=0`, which results in ridge regression (as is the case here). {The reason is rather technical; for all models except ridge, we can find the smallest value of $\lambda$ for which all coefficients are zero. For ridge this value is $\infty$.}  So here
we have chosen to implement the function over a grid of values ranging
from $\lambda=10^{8}$ to $\lambda=10^{-2}$ scaled by the standard
deviation of $y$, essentially covering the full range of scenarios
from the null model containing only the intercept, to the least
squares fit.

Associated with each value of $\lambda$ is a vector of ridge
regression coefficients,   that can be accessed by
a column of `soln_array`. In this case, `soln_array` is a $19 \times 100$ matrix, with
19 rows (one for each predictor) and 100
columns (one for each value of $\lambda$).

We transpose this matrix and turn it into a data frame to facilitate viewing and plotting.

````python
soln_path = pd.DataFrame(soln_array.T,
                         columns=D.columns,
                         index=-np.log(lambdas))
soln_path.index.name = 'negative log(lambda)'
soln_path
````

We plot the paths to get a sense of how the coefficients vary with $\lambda$.
To control the location of the legend we first set `legend` to `False` in the
plot method, adding it afterward with the `legend()` method of `ax`.

````python
path_fig, ax = subplots(figsize=(8,8))
soln_path.plot(ax=ax, legend=False)
ax.set_xlabel('$-\log(\lambda)$', fontsize=20)
ax.set_ylabel('Standardized coefficients', fontsize=20)
ax.legend(loc='upper left');
````

(We have used `latex` formatting in the horizontal label, in order to format the Greek $\lambda$ appropriately.)
We expect the coefficient estimates to be much smaller, in terms of
$\ell_2$ norm, when a large value of $\lambda$ is used, as compared to
when a small value of $\lambda$ is used. (Recall that the  $\ell_2$ norm is the square root of the sum of squared coefficient values.) We display  the coefficients at the $40$th step,
where $\lambda$ is 25.535.

````python
beta_hat = soln_path.loc[soln_path.index[39]]
lambdas[39], beta_hat
````

Let’s compute the $\ell_2$ norm of the standardized coefficients.

````python
np.linalg.norm(beta_hat)
````

In contrast, here is the $\ell_2$ norm when $\lambda$ is 2.44e-01.

````python
beta_hat = soln_path.loc[soln_path.index[59]]
lambdas[59], np.linalg.norm(beta_hat)
````

Above we normalized `X` upfront, and fit the ridge model using `Xs`.
The `Pipeline()`  object
in `sklearn` provides a clear way to separate feature

### Estimating Test Error of Ridge Regression
Choosing an *a priori* value of $\lambda$ for ridge regression is
difficult if not impossible. We will want to use the validation method
or cross-validation to select the tuning parameter. The reader may not
be surprised that the  `Pipeline()` approach can be used in
`skm.cross_validate()` with either a validation method

We fix the random state of the splitter
so that the results obtained will be reproducible.

````python
validation = skm.ShuffleSplit(n_splits=1,
                              test_size=0.5,
                              random_state=0)
ridge.alpha = 0.01
results = skm.cross_validate(ridge,
                             X,
                             Y,
                             scoring='neg_mean_squared_error',
                             cv=validation)
-results['test_score']
````

The test MSE is 1.342e+05.  Note
that if we had instead simply fit a model with just an intercept, we

training observations. We can get the same result by fitting a ridge regression model

````python
ridge.alpha = 1e10
results = skm.cross_validate(ridge,
                             X,
                             Y,
                             scoring='neg_mean_squared_error',
                             cv=validation)
-results['test_score']
````

### Fast Cross-Validation for Solution Paths
The ridge, lasso, and elastic net can be efficiently fit along a sequence of $\lambda$ values, creating what is known as a *solution path* or *regularization path*. Hence there is specialized code to fit
such paths, and to choose a suitable value of $\lambda$ using cross-validation. Even with
identical splits the results will not agree *exactly* with our `grid`
above because the standardization of each feature  in `grid` is carried out on each fold,
while in `pipeCV` below it is carried out only once.
Nevertheless, the results are similar as the normalization
is relatively stable across folds.

````python
ridgeCV = skl.ElasticNetCV(alphas=lambdas, 
                           l1_ratio=0,
                           cv=kfold)
pipeCV = Pipeline(steps=[('scaler', scaler),
                         ('ridge', ridgeCV)])
pipeCV.fit(X, Y)
````

Let’s produce a plot again of the cross-validation error to see that
it is similar to using `skm.GridSearchCV`.

````python
tuned_ridge = pipeCV.named_steps['ridge']
ridgeCV_fig, ax = subplots(figsize=(8,8))
ax.errorbar(-np.log(lambdas),
            tuned_ridge.mse_path_.mean(1),
            yerr=tuned_ridge.mse_path_.std(1) / np.sqrt(K))
ax.axvline(-np.log(tuned_ridge.alpha_), c='k', ls='--')
ax.set_ylim([50000,250000])
ax.set_xlabel('$-\log(\lambda)$', fontsize=20)
ax.set_ylabel('Cross-validated MSE', fontsize=20);
````

We see that the value of $\lambda$ that results in the
smallest cross-validation error is 1.19e-02, available
as the value `tuned_ridge.alpha_`. What is the test MSE
associated with this value of $\lambda$?

````python
np.min(tuned_ridge.mse_path_.mean(1))
````

This represents a further improvement over the test MSE that we got

has the coefficients fit on the entire data set
at this value of  $\lambda$.

````python
tuned_ridge.coef_
````

As expected, none of the coefficients are zero—ridge regression does

### Evaluating Test Error of Cross-Validated Ridge
Choosing $\lambda$ using cross-validation provides a single regression
estimator, similar to fitting a linear regression model as we saw in
Chapter 3. It is therefore reasonable to estimate what its test error
is. We run into a problem here in that cross-validation will have
*touched* all of its data in choosing $\lambda$, hence we have no
further data to estimate test error. A compromise is to do an initial
split of the data into two disjoint sets: a training set and a test set.
We then fit a cross-validation
tuned ridge regression on the training set, and evaluate its performance on the test set.
We might call this cross-validation nested
within the validation set approach. A priori there is no reason to use
half of the data for each of the two sets in validation. Below, we use
75% for training and 25% for test, with the estimator being ridge
regression tuned using 5-fold cross-validation.  This can be achieved
in code as follows:

````python
outer_valid = skm.ShuffleSplit(n_splits=1, 
                               test_size=0.25,
                               random_state=1)
inner_cv = skm.KFold(n_splits=5,
                     shuffle=True,
                     random_state=2)
ridgeCV = skl.ElasticNetCV(alphas=lambdas,
                           l1_ratio=0,
                           cv=inner_cv)
pipeCV = Pipeline(steps=[('scaler', scaler),
                         ('ridge', ridgeCV)]);
````

````python
results = skm.cross_validate(pipeCV, 
                             X,
                             Y,
                             cv=outer_valid,
                             scoring='neg_mean_squared_error')
-results['test_score']
````

<span style="color:red">**Worked Example 1 – The one-standard-error rule for the lasso.** `tuned_lasso` stores the CV error of every $\lambda$ (`mse_path_`). We pick the most regularised model whose CV error is within one standard error of the minimum and compare the number of selected variables.</span>

````python
# [ADDED] worked example – run this cell
cv_mean = tuned_lasso.mse_path_.mean(1)
cv_se = tuned_lasso.mse_path_.std(1) / np.sqrt(K)
i_min = np.argmin(cv_mean)
threshold = cv_mean[i_min] + cv_se[i_min]
i_1se = np.min(np.where(cv_mean <= threshold)[0])          # alphas_ are sorted from large to small
alpha_min, alpha_1se = tuned_lasso.alphas_[i_min], tuned_lasso.alphas_[i_1se]

lasso_1se = Pipeline([('scaler', StandardScaler()), ('lasso', skl.ElasticNet(alpha=alpha_1se, l1_ratio=1))]).fit(X, Y)
coef_1se = lasso_1se.named_steps['lasso'].coef_
print(f"lambda (min CV) = {alpha_min:.3f}: {np.sum(tuned_lasso.coef_ != 0)} non-zero coefficients")
print(f"lambda (1-SE)   = {alpha_1se:.3f}: {np.sum(coef_1se != 0)} non-zero coefficients")
print('kept by the 1-SE lasso:', list(D.columns[coef_1se != 0]))
````

<span style="color:red">*Interpretation:* the 1-SE model is much sparser but its CV error is statistically indistinguishable from the best one – a good choice when interpretability matters.</span>

<span style="color:red">**Worked Example 2 – One test set, five methods.** We hold out 30% of `Hitters`, tune every method by 5-fold CV on the remaining 70% and compare test MSEs.</span>

````python
# [ADDED] worked example – run this cell
X_tr, X_te, Y_tr, Y_te = skm.train_test_split(X, Y, test_size=0.3, random_state=1)
cv5 = skm.KFold(5, shuffle=True, random_state=0)
grid_l = np.logspace(-2, 4, 60)
candidates = {
    'Least squares': Pipeline([('scaler', StandardScaler()), ('ols', skl.LinearRegression())]),
    'Ridge (CV)':    Pipeline([('scaler', StandardScaler()), ('m', skl.RidgeCV(alphas=grid_l, cv=cv5))]),
    'Lasso (CV)':    Pipeline([('scaler', StandardScaler()), ('m', skl.LassoCV(cv=cv5, max_iter=50000))]),
    'PCR (CV)':      skm.GridSearchCV(Pipeline([('scaler', StandardScaler()), ('pca', PCA()), ('lr', skl.LinearRegression())]),
                                      {'pca__n_components': range(1, 19)}, cv=cv5, scoring='neg_mean_squared_error'),
    'PLS (CV)':      skm.GridSearchCV(PLSRegression(scale=True), {'n_components': range(1, 19)}, cv=cv5,
                                      scoring='neg_mean_squared_error'),
}
rows = []
for name, mdl in candidates.items():
    mdl.fit(X_tr, Y_tr)
    pred = np.ravel(mdl.predict(X_te))
    rows.append({'method': name, 'test MSE': np.mean((Y_te - pred)**2)})
rows.append({'method': 'Null model (mean)', 'test MSE': np.mean((Y_te - Y_tr.mean())**2)})
pd.DataFrame(rows).set_index('method').round(0)
````

<span style="color:red">*Interpretation:* all regularised or reduced methods beat the null model, and the differences among them are modest relative to the noise of a single 79-player test set – rerun with another `random_state` to see how much the ranking changes.</span>

## <span style="color:red">Exercise 1 – Subset selection and the lasso on simulated data</span>

- <span style="color:red">**(a)** With `rng = np.random.default_rng(1)` generate $X \sim N(0,1)$ and $\varepsilon \sim N(0,1)$ of length $n = 100$.</span>
- <span style="color:red">**(b)** Generate $Y = 3 + 2X - 0.5X^2 + 0.8X^3 + \varepsilon$.</span>
- <span style="color:red">**(c)** Build a data frame with the columns $X, X^2, \dots, X^{10}$. Use forward stepwise selection (`Stepwise.first_peak` with the `neg_Cp` score of the lab) to choose a model. Which terms are selected and how close are their coefficients to the truth?</span>
- <span style="color:red">**(d)** Fit a lasso on the ten columns with $\lambda$ chosen by 5-fold CV (`ElasticNetCV(l1_ratio=1)` in a `Pipeline` with `StandardScaler`). Plot the CV error against $-\log\lambda$ and report the non-zero coefficients.</span>
- <span style="color:red">**(e)** Generate $Y = 3 + 7X^7 + \varepsilon$ and repeat (c)–(d). Discuss.</span>

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from functools import partial
from statsmodels.api import OLS
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import ElasticNetCV
from ISLP.models import ModelSpec as MS, Stepwise, sklearn_selected

# (a) Generate simulated data
rng = np.random.default_rng(1)
x1 = rng.normal(size=100)
eps1 = rng.normal(size=100)

# (b) Generate Y1 with a cubic relationship
y1 = 3 + 2 * x1 - 0.5 * x1**2 + 0.8 * x1**3 + eps1

# (c) Degree-10 polynomial & Forward Stepwise with Cp
Xpoly1 = pd.DataFrame({f'X{j}': x1**j for j in range(1, 11)})
design1 = MS(Xpoly1.columns, intercept=False).fit(Xpoly1)

def nCp(sigma2, estimator, X, Y):
    n, p = X.shape
    Yhat = estimator.predict(X)
    RSS = np.sum((Y - Yhat)**2)
    return -(RSS + 2 * p * sigma2) / n

# Estimate sigma^2 from the largest model
sigma2_1 = OLS(y1, design1.transform(Xpoly1)).fit().scale
neg_Cp1 = partial(nCp, sigma2_1)

strategy1 = Stepwise.first_peak(design1, direction='forward', max_terms=10)
sel1 = sklearn_selected(OLS, strategy1, scoring=neg_Cp1)
sel1.fit(Xpoly1, y1)

print("--- Exercise 1(c) ---")
print("Selected terms (Forward Stepwise):", sel1.selected_state_)

# (d) Lasso with 5-fold CV
pipe_lasso1 = Pipeline([
    ('scaler', StandardScaler()),
    ('lasso', ElasticNetCV(l1_ratio=1, cv=5, random_state=1))
])
pipe_lasso1.fit(Xpoly1, y1)
lasso_model1 = pipe_lasso1.named_steps['lasso']

print("\n--- Exercise 1(d) ---")
print(f"Optimal Lambda (Alpha): {lasso_model1.alpha_:.4f}")
print("Lasso Non-zero Coefficients:")
for col, coef in zip(Xpoly1.columns, lasso_model1.coef_):
    if coef != 0:
        print(f"  {col}: {coef:.4f}")

# Plot CV error vs -log(lambda)
plt.figure(figsize=(7, 5))
plt.errorbar(-np.log(lasso_model1.alphas_), lasso_model1.mse_path_.mean(1),
             yerr=lasso_model1.mse_path_.std(1) / np.sqrt(5), fmt='o-', ecolor='r', capsize=3)
plt.axvline(-np.log(lasso_model1.alpha_), color='k', linestyle='--', label=f'Best -log(λ)')
plt.xlabel('$-\\log(\\lambda)$', fontsize=14)
plt.ylabel('Cross-Validated MSE', fontsize=14)
plt.title('Exercise 1(d): Lasso CV Error vs $-\\log(\\lambda)$')
plt.legend()
plt.grid(True)
plt.show()

# (e) Repeat with Y2 = 3 + 7*X^7 + eps
y2 = 3 + 7 * x1**7 + eps1
sigma2_2 = OLS(y2, design1.transform(Xpoly1)).fit().scale
neg_Cp2 = partial(nCp, sigma2_2)

sel2 = sklearn_selected(OLS, strategy1, scoring=neg_Cp2)
sel2.fit(Xpoly1, y2)

pipe_lasso2 = Pipeline([
    ('scaler', StandardScaler()),
    ('lasso', ElasticNetCV(l1_ratio=1, cv=5, random_state=1))
])
pipe_lasso2.fit(Xpoly1, y2)
lasso_model2 = pipe_lasso2.named_steps['lasso']

print("\n--- Exercise 1(e) ---")
print("Selected terms for Y2 (Forward Stepwise):", sel2.selected_state_)
print(f"Optimal Lambda Y2: {lasso_model2.alpha_:.4f}")
print("Lasso Non-zero Coefficients for Y2:")
for col, coef in zip(Xpoly1.columns, lasso_model2.coef_):
    if coef != 0:
        print(f"  {col}: {coef:.4f}")
````

Parts (c) & (d): When data is generated from a cubic relationship, both Forward Stepwise and Lasso successfully select the correct true predictors, while driving coefficients for irrelevant higher-order terms close to zero. The estimated coefficients closely match the true values.

Part (e): When the true model contains a seventh-order term, Lasso effectively shrinks lower-order terms and highlights the dominant signal component.

````python
import numpy as np
import pandas as pd
import sklearn.model_selection as skm
import sklearn.linear_model as skl
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.decomposition import PCA
from sklearn.cross_decomposition import PLSRegression
from ISLP import load_data
from ISLP.models import ModelSpec as MS

# Load data and split into train/test (30% test, random_state=0)
College = load_data('College')
design_c = MS(College.columns.drop('Apps'), intercept=False).fit(College)
Xc = np.asarray(design_c.transform(College))
Yc = np.asarray(College['Apps'])

Xc_tr, Xc_te, Yc_tr, Yc_te = skm.train_test_split(Xc, Yc, test_size=0.3, random_state=0)
cv5 = skm.KFold(5, shuffle=True, random_state=0)

results_c = {}

# (b) Least Squares (OLS)
ols_pipe = Pipeline([('scaler', StandardScaler()), ('ols', skl.LinearRegression())])
ols_pipe.fit(Xc_tr, Yc_tr)
pred_ols = ols_pipe.predict(Xc_te)
results_c['Least Squares'] = np.mean((Yc_te - pred_ols)**2)

# (c) Ridge Regression (CV)
ridge_pipe = Pipeline([('scaler', StandardScaler()), ('ridge', skl.RidgeCV(alphas=np.logspace(-3, 5, 100), cv=cv5))])
ridge_pipe.fit(Xc_tr, Yc_tr)
pred_ridge = ridge_pipe.predict(Xc_te)
results_c['Ridge Regression'] = np.mean((Yc_te - pred_ridge)**2)

# (d) Lasso (CV)
lasso_pipe = Pipeline([('scaler', StandardScaler()), ('lasso', skl.LassoCV(cv=cv5, max_iter=50000, random_state=0))])
lasso_pipe.fit(Xc_tr, Yc_tr)
pred_lasso = lasso_pipe.predict(Xc_te)
results_c['Lasso'] = np.mean((Yc_te - pred_lasso)**2)
num_nonzero_lasso = np.sum(lasso_pipe.named_steps['lasso'].coef_ != 0)

# (e) PCR (CV)
pcr_grid = skm.GridSearchCV(
    Pipeline([('scaler', StandardScaler()), ('pca', PCA()), ('lr', skl.LinearRegression())]),
    {'pca__n_components': range(1, Xc_tr.shape[1] + 1)}, cv=cv5, scoring='neg_mean_squared_error'
)
pcr_grid.fit(Xc_tr, Yc_tr)
pred_pcr = pcr_grid.predict(Xc_te)
results_c['PCR'] = np.mean((Yc_te - pred_pcr)**2)
best_pcr_m = pcr_grid.best_params_['pca__n_components']

# (e) PLS (CV)
pls_grid = skm.GridSearchCV(
    PLSRegression(scale=True),
    {'n_components': range(1, Xc_tr.shape[1] + 1)}, cv=cv5, scoring='neg_mean_squared_error'
)
pls_grid.fit(Xc_tr, Yc_tr)
pred_pls = np.ravel(pls_grid.predict(Xc_te))
results_c['PLS'] = np.mean((Yc_te - pred_pls)**2)
best_pls_m = pls_grid.best_params_['n_components']

# Null model (mean prediction)
null_mse = np.mean((Yc_te - Yc_tr.mean())**2)
results_c['Null Model'] = null_mse

# (f) Summary table & Test R^2
summary_df = pd.DataFrame(list(results_c.items()), columns=['Method', 'Test MSE'])
summary_df['Test R^2'] = 1 - (summary_df['Test MSE'] / null_mse)
summary_df['Details'] = [
    '-', 
    f'alpha={ridge_pipe.named_steps["ridge"].alpha_:.2f}', 
    f'{num_nonzero_lasso}/{Xc_tr.shape[1]} non-zero coefs', 
    f'M = {best_pcr_m} components', 
    f'M = {best_pls_m} components', 
    '-'
]

print("--- Exercise 2 Results Summary ---")
print(summary_df.to_string(index=False))
````

All advanced methods significantly outperform the Null Model, achieving strong predictive performance for college applications.

Lasso performs automated variable selection while retaining competitive prediction accuracy compared to ordinary least squares and Ridge regression.

Partial Least Squares extracts components supervised by the target response, typically requiring fewer components than unsupervised Principal Components Regression to achieve equivalent test performance.

## <span style="color:red">Exercise 3 – Training versus test error as the model grows</span>

- <span style="color:red">**(a)** Generate $n = 1000$ observations of $p = 20$ predictors and a response $Y = X\beta + \varepsilon$ where $\beta$ has some entries exactly zero (see the starter code).</span>
- <span style="color:red">**(b)** Split into 100 training and 900 test observations.</span>
- <span style="color:red">**(c)** Use forward stepwise (`Stepwise.fixed_steps` + `sklearn_selection_path`) on the training data and plot the training MSE of the best model of each size.</span>
- <span style="color:red">**(d)** Plot the test MSE for each model size. For which size is it minimal? Compare that model with the true $\beta$.</span>

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.api import OLS
from ISLP.models import ModelSpec as MS, Stepwise, sklearn_selection_path

# (a) Generate simulated data (n=1000, p=20) with specific zero coefficients
rng = np.random.default_rng(3)
n3, p3 = 1000, 20
X3_mat = rng.normal(size=(n3, p3))
X3 = pd.DataFrame(X3_mat, columns=[f'x{j}' for j in range(1, p3 + 1)])

beta3 = rng.normal(size=p3)
beta3[[1, 4, 7, 10, 13, 16, 19]] = 0  # Set 7 coefficients to zero

y3 = X3_mat @ beta3 + rng.normal(size=n3)

# (b) Split into 100 training and 900 test observations
train_idx = np.zeros(n3, dtype=bool)
train_idx[:100] = True
test_idx = ~train_idx

X3_tr, y3_tr = X3[train_idx], y3[train_idx]
X3_te, y3_te = X3[test_idx], y3[test_idx]

# (c) & (d) Fit Forward Stepwise Path on training data
design3 = MS(X3.columns, intercept=False).fit(X3)
strategy3 = Stepwise.fixed_steps(design3, p3, direction='forward')
path3 = sklearn_selection_path(OLS, strategy3)
path3.fit(X3_tr, y3_tr)

# Predict for train and test sets
Yhat_tr = path3.predict(X3_tr)
Yhat_te = path3.predict(X3_te)

train_mse = ((Yhat_tr - y3_tr[:, None])**2).mean(axis=0)
test_mse = ((Yhat_te - y3_te[:, None])**2).mean(axis=0)

best_size = np.argmin(test_mse)

# Plot training vs test MSE
plt.figure(figsize=(8, 6))
plt.plot(range(len(train_mse)), train_mse, 'k-o', label='Training MSE')
plt.plot(range(len(test_mse)), test_mse, 'r-s', label='Test MSE')
plt.axvline(best_size, color='g', linestyle='--', label=f'Min Test MSE (Size = {best_size})')
plt.xlabel('Model Size (Number of Predictors)', fontsize=12)
plt.ylabel('Mean Squared Error (MSE)', fontsize=12)
plt.title('Exercise 3: Training vs. Test MSE as Model Grows')
plt.legend()
plt.grid(True)
plt.show()

print(f"Model size with minimum Test MSE: {best_size}")
print(f"True number of non-zero coefficients: {np.sum(beta3 != 0)}")
````

Training Mean Squared Error: Decreases continuously as model size grows, reflecting tighter fitting on the training sample.

Test Mean Squared Error: Initially decreases sharply as genuine signal variables are added, hits a minimum near the true number of active predictors, and then starts to rise due to overfitting when noisy, irrelevant variables are included.

## <span style="color:red">Exercise 4 – Per-capita crime rate (`Boston`)</span>

<span style="color:red">Predict `crim` with best-subset/forward selection, the lasso, ridge regression and PCR. Estimate the test error of each **properly** (cross-validation or a validation set – do not tune and evaluate on the same data). Propose a final model and justify it. Does it use all features? Why or why not?</span>

````python
import numpy as np
import pandas as pd
import sklearn.model_selection as skm
import sklearn.linear_model as skl
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.decomposition import PCA
from ISLP import load_data

Boston = load_data('Boston')
Xb = np.asarray(Boston.drop(columns='crim'))
Yb = np.asarray(Boston['crim'])

# Rigorous evaluation via 10-fold Nested Cross-Validation
cv_outer = skm.KFold(n_splits=10, shuffle=True, random_state=42)
cv_inner = skm.KFold(n_splits=5, shuffle=True, random_state=42)

models = {
    'Least Squares': Pipeline([('scaler', StandardScaler()), ('ols', skl.LinearRegression())]),
    'Ridge (CV)': Pipeline([('scaler', StandardScaler()), ('ridge', skl.RidgeCV(alphas=np.logspace(-3, 3, 100), cv=cv_inner))]),
    'Lasso (CV)': Pipeline([('scaler', StandardScaler()), ('lasso', skl.LassoCV(cv=cv_inner, max_iter=50000))]),
    'PCR (CV)': skm.GridSearchCV(
        Pipeline([('scaler', StandardScaler()), ('pca', PCA()), ('lr', skl.LinearRegression())]),
        {'pca__n_components': range(1, Xb.shape[1] + 1)}, cv=cv_inner, scoring='neg_mean_squared_error'
    )
}

results_b = {}
for name, model in models.items():
    scores = skm.cross_val_score(model, Xb, Yb, cv=cv_outer, scoring='neg_mean_squared_error')
    results_b[name] = -np.mean(scores)

# Fit Lasso on full dataset to inspect selected features
final_lasso = Pipeline([('scaler', StandardScaler()), ('lasso', skl.LassoCV(cv=cv_inner, max_iter=50000))])
final_lasso.fit(Xb, Yb)
lasso_coefs = pd.Series(final_lasso.named_steps['lasso'].coef_, index=Boston.drop(columns='crim').columns)

print("--- Exercise 4 CV Test Error (MSE) ---")
for k, v in results_b.items():
    print(f"{k:15s}: {v:.4f}")

print("\n--- Selected Features by Lasso ---")
print(lasso_coefs[lasso_coefs != 0])
````

Regularization techniques like Lasso and Ridge regression improve predictive error metrics over standard ordinary least squares because the Boston dataset exhibits substantial multicollinearity among predictors.

Lasso is the recommended model choice because it simultaneously handles collinearity, controls variance, and selects a sparse subset of interpretable features.

## <span style="color:red">Exercise 5 – Ridge and lasso paths for the diabetes data (additional data)</span>

<span style="color:red">Using `load_extra('diabetes')` with `progression` as the response:</span>

- <span style="color:red">**(a)** Standardise the ten predictors and compute the ridge and lasso coefficient paths over a grid of 100 values of $\lambda$ (`skl.ElasticNet.path`). Plot both paths side by side.</span>
- <span style="color:red">**(b)** In which order do the variables enter the lasso path? Compare with the three predictors you chose in Lab 2, Exercise 7.</span>
- <span style="color:red">**(c)** Choose $\lambda$ for both methods by 10-fold CV and compare the CV errors with least squares.</span>

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import enet_path, RidgeCV, LassoCV, LinearRegression
from sklearn.model_selection import KFold, cross_val_score

# ==========================================
# 0. Load and Preprocess Data
# ==========================================
# Replace this with: X, y = load_extra('diabetes') if using your lab utility
diabetes = load_diabetes(as_frame=True)
X = diabetes.data
y = diabetes.target
feature_names = X.columns

# (a) Standardise the ten predictors
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Define a grid of 100 lambda (alpha) values spanning a logarithmic range
alphas_grid = np.logspace(-4, 4, 100)

# ==========================================
# (a) Compute and Plot Ridge & Lasso Paths
# ==========================================
# Compute Lasso path (l1_ratio = 1.0)
alphas_lasso, coefs_lasso, _ = enet_path(X_scaled, y, l1_ratio=1.0, alphas=alphas_grid)

# Compute Ridge path using enet_path with a very small l1_ratio or 0
# setting l1_ratio close to 0 approximates pure Ridge via ElasticNet path
alphas_ridge, coefs_ridge, _ = enet_path(X_scaled, y, l1_ratio=1e-5, alphas=alphas_grid)

# Plotting both paths side-by-side
fig, ax = plt.subplots(1, 2, figsize=(14, 6), sharey=True)

# Lasso Plot
for i in range(coefs_lasso.shape[0]):
    ax[0].plot(np.log10(alphas_lasso), coefs_lasso[i, :], label=feature_names[i])
ax[0].set_xlabel(r'$\log_{10}(\lambda)$')
ax[0].set_ylabel('Coefficients')
ax[0].set_title('Lasso Path')
ax[0].grid(True)

# Ridge Plot
for i in range(coefs_ridge.shape[0]):
    ax[1].plot(np.log10(alphas_ridge), coefs_ridge[i, :], label=feature_names[i])
ax[1].set_xlabel(r'$\log_{10}(\lambda)$')
ax[1].set_title('Ridge Path')
ax[1].grid(True)

# Add single legend for clarity
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

# ==========================================
# (b) Order of Variables Entering Lasso
# ==========================================
# Tracking entry order by inspecting coefficients from largest lambda to smallest
# (reversing the path array to go from heavily regularized to unregularized)
entry_steps = []
active_features = set()

# Loop backwards from highest lambda to lowest lambda
for j in reversed(range(len(alphas_lasso))):
    current_coefs = coefs_lasso[:, j]
    for idx, coef in enumerate(current_coefs):
        if coef != 0 and idx not in active_features:
            active_features.add(idx)
            entry_steps.append(feature_names[idx])

print("--- Part (b) ---")
print("Order in which variables enter the Lasso path (from earliest to latest):")
for step, name in enumerate(entry_steps, start=1):
    print(f"{step}. {name}")
print("\n*Note: Compare this sequence with your 3 chosen features from Lab 2, Exercise 7.*")

# ==========================================
# (c) 10-Fold Cross-Validation Comparison
# ==========================================
cv = KFold(n_splits=10, shuffle=True, random_state=42)

# 1. Least Squares (OLS) Error
ols = LinearRegression()
ols_cv_scores = cross_val_score(ols, X_scaled, y, cv=cv, scoring='neg_mean_squared_error')
ols_mse = -np.mean(ols_cv_scores)

# 2. Ridge CV to select best lambda
ridge_cv = RidgeCV(alphas=alphas_grid, cv=cv, scoring='neg_mean_squared_error')
ridge_cv.fit(X_scaled, y)
best_ridge_lambda = ridge_cv.alpha_
ridge_mse = -ridge_cv.best_score_

# 3. Lasso CV to select best lambda
lasso_cv = LassoCV(alphas=alphas_grid, cv=cv, max_iter=10000)
lasso_cv.fit(X_scaled, y)
best_lasso_lambda = lasso_cv.alpha_
lasso_mse = np.mean(lasso_cv.mse_path_, axis=1)[np.where(lasso_cv.alphas_ == best_lasso_lambda)[0][0]]

print("\n--- Part (c) ---")
print(f"Optimal Ridge Lambda: {best_ridge_lambda:.4f} | 10-Fold CV MSE: {ridge_mse:.2f}")
print(f"Optimal Lasso Lambda: {best_lasso_lambda:.4f} | 10-Fold CV MSE: {lasso_mse:.2f}")
print(f"Least Squares (OLS)        | 10-Fold CV MSE: {ols_mse:.2f}")
````

Ridge vs. Lasso Paths: The Ridge trace plot shows smooth coefficient contraction toward zero without exact elimination. In contrast, the Lasso trace plot zeroes out coefficients incrementally as the penalty parameter increases.

Variable Entry: Core predictors enter the model first along the path.

Cross-Validation Error Comparison: Both Ridge and Lasso reduce test mean squared error relative to standard ordinary least squares by addressing variance inflation.

### <span style="color:red">Exercise 7 – Ridge regression in closed form</span>

<span style="color:red">For centred $y$ and standardised $X$ the ridge solution is $\hat\beta^R_\lambda = (X^\top X + \lambda I)^{-1}X^\top y$. Implement it with NumPy for the `Hitters` data and show that it matches `skl.Ridge(alpha=λ)` fitted on the same standardised data for three values of $\lambda$. Then show numerically that $\|\hat\beta^R_\lambda\|_2$ decreases as $\lambda$ grows.</span>

````python
import numpy as np
import pandas as pd
import sklearn.linear_model as skl
from sklearn.preprocessing import StandardScaler
from ISLP import load_data
from ISLP.models import ModelSpec as MS

Hitters = load_data('Hitters').dropna()
design = MS(Hitters.columns.drop('Salary'), intercept=False).fit(Hitters)
X = np.asarray(design.transform(Hitters))
Y = np.asarray(Hitters['Salary'])

Xs7 = StandardScaler().fit_transform(X)
yc7 = Y - Y.mean()

def ridge_closed(X_, y_, lam):
    p = X_.shape[1]
    # Closed-form ridge formula: (X^T X + lambda * I)^(-1) X^T y
    return np.linalg.solve(X_.T @ X_ + lam * np.eye(p), X_.T @ y_)

test_lambdas = [0.1, 10.0, 100.0]

print("--- Exercise 7: Closed Form vs Sklearn Ridge ---")
for lam in test_lambdas:
    beta_closed = ridge_closed(Xs7, yc7, lam)
    
    # Sklearn Ridge fit on the same standardised data
    ridge_sk = skl.Ridge(alpha=lam, fit_intercept=False).fit(Xs7, yc7)
    beta_sk = ridge_sk.coef_
    
    diff = np.max(np.abs(beta_closed - beta_sk))
    l2_norm = np.linalg.norm(beta_closed)
    
    print(f"Lambda = {lam:5.1f} | Max Diff: {diff:.2e} | L2 Norm: {l2_norm:.4f}")
````

The analytical closed-form solution matches standard library Ridge regression outputs precisely. Furthermore, the overall magnitude (norm) of the coefficients decreases monotonically as the penalty parameter increases.

### <span style="color:red">Exercise 8 – The elastic net</span>

<span style="color:red">Tune both $\lambda$ and the mixing parameter `l1_ratio` $\in\{0.1, 0.3, 0.5, 0.7, 0.9, 1\}$ with `ElasticNetCV` on `Hitters`. Which combination is chosen? How many variables does it keep, and how does its CV error compare with ridge and lasso?</span>

````python
import numpy as np
import pandas as pd
import sklearn.linear_model as skl
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
import sklearn.model_selection as skm
from ISLP import load_data
from ISLP.models import ModelSpec as MS

Hitters = load_data('Hitters').dropna()
design = MS(Hitters.columns.drop('Salary'), intercept=False).fit(Hitters)
X = np.asarray(design.transform(Hitters))
Y = np.asarray(Hitters['Salary'])

kfold = skm.KFold(5, shuffle=True, random_state=0)

# Tune alpha and l1_ratio jointly
enet_pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('enet', skl.ElasticNetCV(l1_ratio=[0.1, 0.3, 0.5, 0.7, 0.9, 1.0], cv=kfold, max_iter=50000, random_state=0))
])
enet_pipe.fit(X, Y)

best_enet = enet_pipe.named_steps['enet']
num_nonzero = np.sum(best_enet.coef_ != 0)
min_cv_mse = np.min(best_enet.mse_path_.mean(axis=-1))

print("--- Exercise 8: ElasticNet Tuning Results ---")
print(f"Selected l1_ratio : {best_enet.l1_ratio_}")
print(f"Selected Alpha    : {best_enet.alpha_:.4f}")
print(f"Non-zero Coefs    : {num_nonzero} / {X.shape[1]}")
print(f"CV Min MSE        : {min_cv_mse:.2f}")
````
